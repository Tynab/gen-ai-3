```python
import os

from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
load_dotenv()

llm = ChatOpenAI(
    model_name=os.getenv("OPENAI_MODEL"),
    temperature=1,
    api_key=os.getenv("OPENAI_API_KEY")
)
```

```python
messages = [
    ("system", "Bạn là một trợ lý hữu ích."),
    ("human", "Python là gì?"),
]
```

```python
result = llm.invoke(messages)
result
```

```python
result.content
```

```python
from langchain_core.prompts import ChatPromptTemplate

message = [
    ("system", "Bạn là một trợ lý hữu ích, bạn sẽ trả lời các câu hỏi của tôi bằng {language}."),
    ("human", "Python là gì?"),
]

prompt = ChatPromptTemplate.from_messages(message)
```

```python
chain = prompt | llm
```

```python
result = chain.invoke({
    "language": "Bồ Đào Nha",
    "input": "Python là gì?"
})

result.content
```
